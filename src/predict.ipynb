```python
import pickle
from pathlib import Path
from rdkit import Chem
from rdkit.Chem import MACCSkeys,DataStructs
import numpy as np
import pandas as pd

# Notebook lives in src/; support launching Jupyter from either src/ or the repo root
ROOT = Path.cwd().parent if Path.cwd().name == 'src' else Path.cwd()
DATA_DIR = ROOT / 'data'
MODEL_DIR = ROOT / 'models'

def smiles_to_fingerprint(path):
    df = pd.read_excel(path)
    smiles_list = df['SMILES'] 
    res = []
    for smiles in smiles_list:
        
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            print(f'{smiles} error!')
        # fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius, nBits=n_bits)
        fp = MACCSkeys.GenMACCSKeys(mol)
        arr = np.zeros((1,))
        DataStructs.ConvertToNumpyArray(fp, arr)
        res.append(arr)
    return res

def transfer_(x):
    class_type = {0:'A', 1:'B', 2:'C', 3:'D', 4:'E', 5:'F', 6:'G', 7:'H', 8:'I', 9:'J', 10:'K'}
    return class_type[x]
# smiles_to_fingerprint(DATA_DIR / 'valid_data.xlsx')
```

```python
import sys
sys.path.insert(0, str(ROOT))
from main import load_model_bundle

# 模型是在标准化后的特征上训练的, 必须连同 scaler / label_encoder 一起加载
model, scaler, label_encoder, model_name = load_model_bundle(MODEL_DIR / 'best_model.pkl')
print(f'已加载模型: {model_name}')
```

```python
# X_test  = np.array([[0] * 167])
X_test  = np.array(smiles_to_fingerprint(DATA_DIR / 'valid_data.xlsx'))
# X_test  = smiles_to_fingerprint(DATA_DIR / 'train_data.xlsx')

# 关键: 用训练时同一个 scaler 做变换, 否则树阈值对不上, 预测会退化成多数类
X_test_scaled = scaler.transform(X_test)

# 使用加载的模型进行预测
y_pred = model.predict(X_test_scaled)

# 打印预测结果
print("Predictions:", y_pred)
```
